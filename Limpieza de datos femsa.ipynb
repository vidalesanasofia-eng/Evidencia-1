# Limpieza de datos FEMSA


En este reporte se llevara acabo una limpieza de datos para el socio formador FEMSA de la empresa Spin OXXO. El proposito es elaborar la limpieza de la base de datos para poder hacer un analisis profundo en una base limpia.

### Exportación de datos
Primero se exportan los datos para tener nuestra base lista en python y poder iniciar con la limpieza.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_excel("credito.xlsx")
df.head()
```

### Paso 1:identificación de duplicados, nulos y tipos de datos

El primer paso para una base de datos limpia es la identifiicación de valores duplicados, nulos y tipos de datos. Este procedimiento nos sirve para poder observar que tanta limpieza va a necesitar nuestra base de datos, ya que con esto podemos identificar columnas o valores nulos los cuales no nos sirven para nuestro analisis.


```python
print("La cantidad de filas totales en la base de datos es:",len(df))
```

```python
print("Valores nulos por columna:")
print(df.isnull().sum())
```

```python
print("Valores duplicados por columna:")
print(df.duplicated().sum())
```

```python
print("Tipos de datos por columna:")
print(df.info())
```

### Paso 2:Limpieza de base de datos
En esta sección, removemos todos los datos de las columnas que tengan datos faltantes o datos duplicados, el proposito de este paso es para tener un mejor analisis con datos de calidad, ya que si hay datos que faltan no se puede hacer un analisis completo por lo cual se eliminan y con esto podemos utilizar los datos que si puedan generar valor a nuestro analisis.

```python
df1 = df.dropna(subset=["Estatus_Cobranza", "Solicitud", "Fecha_Deposito", "Fecha_de_primer_pago", "loan_name", "Monto", "Plazo", "Tasa_con_impuesto", "Tasa_sin_impuesto", "Frecuencia", "Cuota", "Cuotas"])
print("Base de datos sin valores nulos")
df
```

```python
df2 = df1.drop_duplicates(subset=["Estatus_Cobranza", "Solicitud", "Fecha_Deposito", "Fecha_de_primer_pago", "loan_name", "Monto", "Plazo", "Tasa_con_impuesto", "Tasa_sin_impuesto", "Frecuencia", "Cuota", "Cuotas"])
print("Base de datos sin valores nulos y duplicados")
df2
```

```python
import pandas as pd
from itertools import combinations

def encontrar_pares_duplicados(df):
    pares = []
    for col1, col2 in combinations(df.columns, 2):
        if df[col1].equals(df[col2]):
            pares.append((col1, col2))
    return pares

pares_duplicados = encontrar_pares_duplicados(df)

for original, duplicada in pares_duplicados:
    print(f"Las columnas duplicadas son: {duplicada} y {original}")
```

Despues de identificar cuales son las columnas duplicadas, se eliminan de nuestra base de datos las que no nos proporcionan datos utiles para nuestro analisis.En este caso "folio 1 " y "Status" fueron las columnas que se removieron de la base de datos debido a que tienen la misma información y hacen que nuestra base este mas limpia y ma simplificada. 

```python
df3 = df2.drop(columns=["folio 1","Status"])
print("Base de datos sin valores nulos, duplicados y sin columnas duplicadas")
df3
```

### Paso 3:Normalización de texto y tipos de datos
En este paso tiene la función de hacer que todos los datos de la columna sean el mismo para que no haya diferencuas entre los tipos de datos y en emomento de hacer el analisis sea un proceso mas fluido.En este caso la base de datos ya tiene los datos por lo cual este paso no es necesario de realizar, solo por cuestiones de formato se remueven los espacios en blanco al inicio/final en columnas de texto. Tambien se muestran todos los titulos de las columnas con mayuscula.


```python
columnas_texto = df3.select_dtypes(include="object").columns
for col in columnas_texto:
    df3[col] = df3[col].str.strip()

df3.columns = df3.columns.str.capitalize()
df3

```

### Paso 4:Variables calculadas
En esta sección se realizaran calculaciones de variables que podrian llegar a ser util para nuestro analisis, de esta forma la hacer el tablero ya tendremos una noción de la información que queremos visualizar.

```python
Tipos_de_prestamo = df3["Loan_name"].value_counts()
Tipos_de_prestamo
```

```python
Plazos_diferentes = df3["Plazo"].value_counts()
Plazos_diferentes
```

```python
Monto_promedio = df3["Monto"].mean()
Monto_minimo = df3["Monto"].min()
Monto_maximo = df3["Monto"].max()
print("El monto promedio es:",Monto_promedio)
print("El monto mínimo es:",Monto_minimo)
print("El monto máximo es:",Monto_maximo)
```
